##Genre 
Find median avg playtime for all genres 

```python
%sql
SELECT COUNT(DISTINCT gm.app_id) AS played_games_with_genre
FROM steam_data.default.gold_games gm
JOIN steam_data.default.game_genres gg ON gm.app_id = gg.app_id
WHERE gm.average_playtime_forever > 0;
```

```python
    %sql
    SELECT g.description, round(avg(gm.average_playtime_forever),2)
    FROM steam_data.default.gold_games gm
    JOIN steam_data.default.game_genres gg ON gm.app_id = gg.app_id
    JOIN steam_data.default.genres g      ON gg.genre_id = g.genre_id
    WHERE gm.average_playtime_forever > 0
    GROUP by g.description
    ;
```
