```python
%sql
CREATE TABLE Department (
    id INT PRIMARY KEY,
    name VARCHAR(50)
);

CREATE TABLE Employee (
    id INT PRIMARY KEY,
    name VARCHAR(50),
    departmentId INT,
    salary INT,
    hire_date DATE,
    FOREIGN KEY (departmentId) REFERENCES Department(id)
);
```

```python
%sql
SELECT DISTINCT e.departmentId
FROM Employee e
WHERE EXISTS (
    SELECT 1
    FROM Employee e1
    WHERE e1.departmentId = e.departmentId
      AND e1.salary > 100000
);
```
